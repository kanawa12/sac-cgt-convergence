```python
_, _, ss_plant, ss_refmodel, symdict = ks.build_ss_plant_and_refmodel(pdim, rdim, "obs", "obs")
# 制御対象と規範モデルのsym StateSpaceの作成 symdictはシンボリック変数のリスト

SMat, sdict = ks.make_smats(pdim, rdim)
# SAC,CGTで出てくるS_x, S_u, K_x, K_uの定義

symdict = symdict | sdict


expr, kas = ks.make_CGTEqs(SMat, ss_plant, ss_plant)
# CGT方程式の作成
#dp(ty:=ks.util_display_cgteq(kas)); print(sp.latex(ty))
ssdict = ks.solve_CGTEq(expr, svals:=symdict["Sx"] + symdict["Su"])
# CGT方程式をS行列について解き、解を辞書の形で返す


exprs2 = ks.elim_zero_row(list(sp.simplify(expr.subs(ssdict))))
print("exprs2:")
dp(exprs2)
dp(exprs3 := [exprs2[0], exprs2[2]])
dp(kvals := tuple(kas[1].blocks[1,0]))
dp(sp.solve(exprs3[1], kvals[0]))

eq, abans, kans = ks.solve_CGTEq2(expr, ssdict, symdict["Ap"], symdict["Bp"], symdict["Kx"], symdict["Ku"])

dp(kans)

# CGT方程式にS行列を代入して消去し、(Ap,Bp), (Kx,Ku) それぞれについてまとめる
resultdict = {
    "plant": ss_plant, "refmodel": ss_refmodel, "SMatrix": SMat, "CGT_expr": expr, "KAS": kas,
    "ssdict": ssdict, "plainEq":eq, "apbpEq": abans, "kxkuEq": kans
}
#dp(resultdict)
```
